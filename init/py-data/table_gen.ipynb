```python
import random
import re

# 1. Generare nume de persoana bazat pe ./nume/ro_nume_familie.txt si ./nume/ro_prenume.txt
def remove_diacritics(text):
   """Remove Romanian diacritics from text using regex"""
   # Map Romanian characters with diacritics to their base forms
   diacritics_map = {
      'ă': 'a', 'Ă': 'A',
      'â': 'a', 'Â': 'A',
      'î': 'i', 'Î': 'I',
      'ș': 's', 'Ș': 'S',
      'ț': 't', 'Ț': 'T',
      'ş': 's', 'Ş': 'S',
   }
   
   # Replace each diacritic character with its base form
   result = text
   for diacritic, base in diacritics_map.items():
      result = result.replace(diacritic, base)
   return result

def load_names_from_file(filepath):
   """Load names from file and remove diacritics"""
   with open(filepath, 'r', encoding='utf-8') as f:
      # Read lines, strip whitespace, remove empty lines, and remove diacritics
      names = [remove_diacritics(line.strip()) for line in f if line.strip()]
   return names

family_names = load_names_from_file('./nume/ro_nume_familie.txt')
first_names = load_names_from_file('./nume/ro_prenume.txt')

def get_random_ro_name():
   """Generate a random Romanian name in format: FIRST_NAME FAMILY_NAME"""
   first_name = random.choice(first_names)
   family_name = random.choice(family_names)
   return f"{first_name} {family_name}"
```
